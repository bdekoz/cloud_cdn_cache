```python
from pathlib import Path
import json
from collections import Counter
ROOT = Path('/home/bkoz/src/cloud_cdn_cache.bkoz')
manifest = json.loads((ROOT / 'data/manifest.20261021.json').read_text())
print({k: manifest[k] for k in ['record_count', 'layer_count', 'infrastructure_evidence_record_count', 'geographic_footprint_view_count']})
```

```python
review = json.loads((ROOT / 'data/microsoft/enrichment-review.20261021.json').read_text())
print(json.dumps({k: review[k] for k in ['location_records_by_service', 'assertions_by_category', 'locality_references']}, indent=2))
assert review['unique_physical_datacenter_count'] is None
assert review['ownership_inventory_complete'] is False
```

```python
infra = json.loads((ROOT / 'data/infrastructure/microsoft-evidence.20261021.json').read_text())
view = json.loads((ROOT / 'data/microsoft/provider-localities.20261021.geojson').read_text())
loughton = next(r for r in infra['records'] if r['name'] == 'Loughton')
assert loughton['properties']['map_eligible'] is False
assert all(loughton['record_id'] not in f['properties']['source_claim_ids'] for f in view['features'])
assert all(f['properties']['physical_site_confirmed'] is False for f in view['features'])
print('Passed: no invented property count, no Loughton locality, approximate references retain zero facility weight.')
```
